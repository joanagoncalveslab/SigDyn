```python
# =========================
# Imports
# =========================
from ete3 import Tree
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os


# =========================
# Global plotting settings
# =========================
plt.rcParams['pdf.fonttype'] = 42
plt.rcParams['ps.fonttype'] = 42
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = 'Arial'


# =========================
# Helper functions
# =========================
def extract_leaf_id(leaf):
    """Consistent extraction of sample ID from ETE leaf."""
    return str(leaf)[3:41]


# =========================
# Load tree
# =========================
TREE_PATH = "../rerooted"

with open(TREE_PATH, "r") as f:
    newick_tree = f.read()

tree = Tree(newick_tree)
root = tree.get_tree_root()


# =========================
# Distances from root
# =========================
distances_from_root = {
    extract_leaf_id(leaf): root.get_distance(leaf, topology_only=False)
    for leaf in tree.iter_leaves()
}


# =========================
# Load signature data
# =========================
SIGNATURE_PATH = (
    "../just_leaves/example_vcf_per_cell/Assignment_Solution/"
    "Activities/Assignment_Solution_Activities.txt"
)

signatures_df = pd.read_csv(SIGNATURE_PATH, sep="\t")

# Drop all-zero signatures
signatures_df = signatures_df.loc[:, (signatures_df != 0).any(axis=0)]

# Index by sample
signatures_df = signatures_df.set_index("Samples")


# =========================
# Combine signatures (consistent naming)
# =========================
signatures_df["SBS5_SBS40c"] = signatures_df["SBS5"] + signatures_df["SBS40c"]
signatures_df["SBS12_SBS26"] = signatures_df["SBS12"] + signatures_df["SBS26"]


# =========================
# Keep only samples present in tree
# =========================
common_leaves = [
    leaf_id for leaf_id in distances_from_root
    if leaf_id in signatures_df.index
]

signatures_df = signatures_df.loc[common_leaves]


# =========================
# Add distance from root
# =========================
signatures_df["distance_from_root"] = [
    distances_from_root[sample] for sample in signatures_df.index
]


# =========================
# Output directories
# =========================
RESULTS_DIR = "results"
FIGURES_DIR = os.path.join(RESULTS_DIR, "figures")

os.makedirs(RESULTS_DIR, exist_ok=True)
os.makedirs(FIGURES_DIR, exist_ok=True)


# =========================
# Signatures to process
# =========================
SIGNATURES = [
    "SBS12_SBS26",
    "SBS5_SBS40c",
    "SBS1",
]


# =========================
# Loop over all signatures
# =========================
for sig in SIGNATURES:
    if sig not in signatures_df.columns:
        print(f"Skipping {sig}: not found")
        continue

    # ---------- Save results table ----------
    out_df = signatures_df[[sig, "distance_from_root"]].copy()
    out_df = out_df.rename(columns={sig: sig})


    out_path = os.path.join(RESULTS_DIR, f"{sig}_results.tsv")
    out_df.to_csv(out_path, sep="\t")

    # ---------- Plot heatmap ----------
    VMIN = 0
    VMAX = signatures_df[SIGNATURES].values.max()

    plot_df = out_df[[sig]]

    fig, ax = plt.subplots(figsize=(4, 20))

    heatmap = sns.heatmap(
        plot_df,
        yticklabels=False,
        ax=ax,
        #vmin=VMIN,
        #vmax=VMAX,
        cbar_kws={"label": sig}
    )

    # Colorbar formatting
    cbar = heatmap.collections[0].colorbar
    cbar.ax.tick_params(labelsize=30)
    cbar.set_label("Signature Exposure", fontsize=30)

    # Axis formatting
    ax.set_xlabel("")
    ax.set_ylabel("")
    ax.tick_params(axis="x", labelsize=20)

    plt.tight_layout()

    fig_path = os.path.join(FIGURES_DIR, f"{sig}_heatmap.pdf")
    plt.savefig(fig_path, format="pdf")
    plt.close()

    print(f"Saved results and figure for {sig}")
```
